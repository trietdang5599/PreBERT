# LLM review preprocessing

This notebook uses `meta-llama/Llama-3.2-3B-Instruct` to keep only review segments that contain sentiment or evaluative information. It calls the unified `python -m exp_llm preprocess` command.

Before running, accept the Llama model license on Hugging Face and authenticate with `huggingface-cli login` or `HF_TOKEN`.

```python
from pathlib import Path

dataset_name = "Small_Toys_and_Games_5"
input_path = Path("data") / f"{dataset_name}.json"
output_path = Path("data") / f"{dataset_name}_llama_filtered.json"
model_id = "meta-llama/Llama-3.2-3B-Instruct"

input_path, output_path
```

## Validate configuration

The dry run validates the JSONL input without loading the model.

```python
import subprocess
import sys

base_command = [
    sys.executable,
    "-m", "exp_llm", "preprocess",
    str(input_path),
    "--output", str(output_path),
    "--model", model_id,
    "--batch-size", "8",
    "--review-batch-size", "64",
]
subprocess.run(base_command + ["--dry-run"], check=True)
```

## Run preprocessing

The output preserves every source field and adds/replaces `filteredReviewText` and `overall_new`. Existing output is protected unless `--overwrite` is supplied. Use `--no-adjust-ratings` if only text filtering is required.

```python
subprocess.run(base_command, check=True)
```

## Inspect output

```python
import json

with output_path.open(encoding="utf-8") as handle:
    preview = [json.loads(next(handle)) for _ in range(5)]

[(row["reviewText"], row["filteredReviewText"], row["overall"], row["overall_new"]) for row in preview]
```
